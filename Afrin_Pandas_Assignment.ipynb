### Q1. Broadcasting with NumPy
Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?



```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

### Explanation:
Shape of a is (3, 1).

Shape of b is (1, 3).

Broadcasting expands the smaller dimensions so the arrays can be added element-wise.

The resulting shape is (3, 3).

If b is np.array([1, 2]), the addition will give a broadcasting error because the dimensions are incompatible.


#### Q2.A CSV file contains commas inside quoted text fields, missing values, and dates in different formats. Explain how incorrect parsing can affect the resulting DataFrame and identify suitable read options or preprocessing steps.
### Given dataset: students.csv
ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80

```python
import pandas as pd
from io import StringIO

data = """ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA"""

df = pd.read_csv(StringIO(data))

print(df)
print(df.dtypes)
```

### 1. Predict how the quoted comma in "Asha, R" will be handled when the CSV file is read.
The comma inside "Asha, R" is enclosed in quotation marks, so Pandas treats it as part of the Name field and does not split it into two columns.
### 2. Identify the potential issues with the Date and Marks columns.
The Date column contains dates in different formats, such as 2026-09-01 and 01/09/2026, so the dates should be converted into a consistent date-time format. The Marks column contains missing values, including a blank value and "NA", which need to be handled carefully.
### 3. Rewrite the code to handle missing marks and convert the date column into a suitable date-time format.

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df["Marks"] = pd.to_numeric(
    df["Marks"],
    errors="coerce"
)

print(df)
print(df.dtypes)
print(df.isnull().sum())
```

### Explanation:
pd.to_datetime() converts the Date column into a date-time format, while pd.to_numeric(errors="coerce") converts Marks into numeric values and changes invalid or missing values into NaN.

### 4. Explain why the value "NA" should be investigated before deciding whether it represents a missing value.
The value "NA" should be investigated because it may have a specific meaning in the dataset. It could represent a missing mark, but it could also indicate another status. Therefore, its meaning should be confirmed before treating it as a missing value.

#### Q3. A salary column contains the values 25000, 30000, 35000, and 10000000. Explain how the extreme value affects the mean compared with the median. Identify which statistic may better represent a typical employee's salary and justify your choice.


```python
import pandas as pd

df = pd.DataFrame({
    'Employee': ['A', 'B', 'C', 'D', 'E'],
    'Salary': [30000, 35000, 40000, 45000, 500000]
})

print("Mean Salary:", df['Salary'].mean())
print("Median Salary:", df['Salary'].median())

highest = df.loc[df['Salary'].idxmax()]
print("\nHighest Salary Record:")
print(highest)
```

### Section D – Scenario-Based
#### Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 


### Step 1: Create and inspect the dataset

```python
import pandas as pd

df = pd.DataFrame({
    "Student_ID": [
        "S101", "S102", "S103", "S104", "S105",
        "S106", "S107", "S108", "S102", "S109",
        "S110", "S111", "S112", "S113", "S114"
    ],
    "Student_Name": [
        "Asha", "Ravi", "Neha", "Karan", "Meera",
        "Arjun", "Priya", "Rahul", "Ravi", "Sneha",
        "Varun", "Diya", "Aman", "Nisha", "Kabir"
    ],
    "Department": [
        "IT", "it", "I.T.", "HR", "HR",
        "Finance", "finance", "IT", "it", "I.T.",
        "HR", "Finance", "IT", "HR", "Finance"
    ],
    "Assessment": [
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python"
    ],
    "Marks": [
        "85", "78", "92", "", "Absent",
        "0", "67", "75", "78", "NA",
        "88 marks", "91", "", "64", "105"
    ]
})

print(df.head())
print(df.info())
print(df.isnull().sum())
```

### Explanation:
This step creates the given student assessment dataset and inspects its structure and missing values.

### Step 2: Standardize Department Names.

```python
df["Department"] = df["Department"].replace({
    "it": "IT",
    "I.T.": "IT",
    "finance": "Finance"
})

print(df["Department"])
```

### Explanation:
This step standardizes inconsistent department names so that IT and Finance have consistent names.

### Step 3 — Identify and remove duplicate records

```python

print("Duplicate records:")
print(df[df.duplicated()])


print("Number of duplicate records:", df.duplicated().sum())


df = df.drop_duplicates()

print("\nData after removing duplicates:")
print(df)
```

### Explanation:

This step identifies exact duplicate records and removes them using drop_duplicates()

### Step 4-Convert the Marks column into numeric format.

```python
# Keep track of absent students before converting Marks
df["Marks_Status"] = df["Marks"].replace({
    "Absent": "Absent",
    "NA": "Not Entered",
    "": "Not Entered"
})

# Remove text " marks"
df["Marks"] = df["Marks"].str.replace(" marks", "", regex=False)

# Convert Marks to numeric
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df)
print("\nData types:")
print(df.dtypes)
```

### Explanation:

This step converts valid marks into numeric values while preserving information about absent and not-entered marks.

### Step 5 — Investigate Missing Marks.

```python
print("Students with missing marks:")
print(df[df["Marks"].isna()][
    ["Student_ID", "Student_Name", "Marks_Status"]
])
```

### Explanation:

This step identifies students with missing numeric marks and distinguishes absent students from marks that were not entered.

### Step 6 — Identify Invalid Marks

```python
# Identify marks outside the valid range
invalid = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print("Invalid marks:")
print(invalid[["Student_ID", "Student_Name", "Marks"]])

print("\nNumber of invalid marks:", len(invalid))

# Treat invalid marks as missing after identifying them
df.loc[(df["Marks"] < 0) | (df["Marks"] > 100), "Marks"] = pd.NA

print("\nData after handling invalid marks:")
print(df)
```

### Explanation:

This step identifies marks outside the valid 0–100 range and treats them as missing after investigation.

### Step 7 — Display the Cleaned Data 

```python
print("Cleaned DataFrame:")
print(df)

print("\nNumber of records removed:", 1)
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:", 1)
```

### Explanation: 
This step displays the cleaned dataset and summarizes the records removed, remaining missing marks, and invalid marks identified.
